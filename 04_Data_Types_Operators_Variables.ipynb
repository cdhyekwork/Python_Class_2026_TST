## 2. Booleans
Operators: `and`, `or`, `not`
Precedence: `not` > `and` > `or`

```python
a = True
b = False
print(a and b)   # conjunction
print(a or b)    # disjunction
print(not a)     # negation
```

```python
# Precedence: a and not b or c  ==  (a and (not b)) or c
a, b, c = True, True, False
print(a and not b or c)
print((a and (not b)) or c)
```

### Extra Stuff

```python
# Look up a character's number
print(ord("1"), ord("9"), ord("A"), ord("a"), ord(" "), ord("Ä"))

# --- Numbers as strings ---
print("10" < "9")          # True:  '1' (49) < '9' (57), decided at first char
print("100" < "2")         # True:  same reason, length doesn't matter
print("10" == "10.0")      # False: different characters
print(int("10") < int("9"))# False: convert first to compare as numbers!

# --- Same start, then a difference ---
print("apple" < "apricot") # True:  'a'='a', 'p'='p', then 'p' (112) < 'r' (114)
print("cat" < "car")       # False: 't' (116) > 'r' (114)

# --- Prefix: shorter wins ---
print("app" < "apple")     # True:  "app" runs out first, a prefix is smaller
print("" < "a")            # True:  the empty string is smaller than everything

# --- Uppercase vs lowercase ---
print("Zebra" < "apple")   # True:  'Z' (90) < 'a' (97), ALL capitals come before lowercase
print("B" < "a")           # True
print("cat" > "Cat")       # True:  'c' (99) > 'C' (67)
print("abc" == "ABC")      # False: comparison is case-sensitive

# --- Digits vs letters, spaces ---
print("9" < "a")           # True:  digits (48-57) come before letters
print(" z" < "a")          # True:  space (32) is smaller than everything visible

# --- Umlauts (relevant for German words!) ---
print("Äpfel" < "Birne")   # False: 'Ä' (196) is larger than all A-z
print("zebra" < "über")    # True:  'z' (122) < 'ü' (252)

# --- Practical consequences ---
print("file10" < "file2")        # True:  why "file10" sorts before "file2"
print("2026-10-27" < "2026-11-01")  # True: ISO dates sort correctly as strings


print(ord("1"), ord("9"), ord("A"), ord("a"), ord(" "), ord("Ä"))
```

## 7. Comparing Floats
⚠ **Floats are imprecise!** Use `abs()` or `round()` when comparing them.

```python
0.1 + 0.2 == 0.3
```

```python
print(0.1 + 0.2)
print(abs((0.1 + 0.2) - 0.3) <= 1e-10)
print(round(0.1 + 0.2, 10) == round(0.3, 10))
```

## 14. Assignments
`var = expr` — the expression is evaluated first, then assigned.
⚠ Don't confuse `=` (assignment) with `==` (comparison)!

```python
x = 1
y = z = 2
u, w = 3, 4
print(x, y, z, u, w)
u, w = w, u          # swap!
print(u, w)
```

Short forms (`++` / `--` don't exist in Python):

```python
x = 5
x += 3
print(x)
x /= 2
print(x)
```

### Tough Stuff

```python
print(6 / 2)        # 3.0   -> '/' always gives float
print(6 // 2)       # 3     -> use // for an int result
print(7 // 2)       # 3     -> floor division cuts off the decimals
print(-7 // 2)      # -4    -> ...by rounding DOWN, not toward 0!
print(2 ** 3)       # 8
print(2 ** -3)      # 0.125 -> negative exponent = 1/8
print(3 * 1.0)      # 3.0   -> one float is enough
print(7 // 2.0)     # 3.0   -> // with a float gives a float (but still floored)
print(7 % 2.0)      # 1.0   -> same for %

print(type(10 / 5))         # <class 'float'>
print(type(10 // 5))        # <class 'int'>
print(int(10 / 5))          # 2  -> convert back if needed
```
