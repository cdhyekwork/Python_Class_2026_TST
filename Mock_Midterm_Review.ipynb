## 2. Valid and Invalid Variable Names
Rules for names:
- only **letters, digits and underscores** `_`
- must **not start with a digit**
- must **not be a keyword** (`if`, `for`, `class`, `while`, …)
- case-sensitive: `total` and `Total` are different names

### 🧪 Try it – which of these are invalid, and why?
`player_1` · `1st_player` · `_hidden` · `best-score` · `levelUp` · `for` · `Total` · `high score` · `str`

Decide first, then run the cell. (You don't need to understand *how* this cell works yet – just look at the results.)

```python
import keyword

names = ["player_1", "1st_player", "_hidden", "best-score", "levelUp",
         "for", "Total", "high score", "str"]

for name in names:
    valid = name.isidentifier() and not keyword.iskeyword(name)
    print(f"{name!r:14} valid: {valid}")
```

```python
# Uncomment one line at a time -> SyntaxError
# 1st_player = 5
# best-score = 10
# for = 3
```

### Valid, but a bad idea: reusing a built-in name
`str` is a valid name – but it's already the name of the `str()` function!

```python
str = "hello"      # allowed ...
print(str)
```

```python
# ⚠️ error on purpose – str() doesn't work any more!
print(str(42))
```

```python
# Run this to repair it: del removes our variable, so str() is back
del str
print(str(42))
```

## 3. Changing Types Step by Step
One value travels through several types. Before each run, predict the **type and value** of every variable.

**Slicing:** `s2[:2]` = characters 0 and 1, `s2[3:]` = everything from index 3 on.

```python
s3 = s2[:2] + s2[3:]          # cut out the character at index 2
print("s3:   ", type(s3), repr(s3))

left = s3[:2]
right = s3[2:]
print("left: ", type(left), repr(left))
print("right:", type(right), repr(right))

total = int(left) + int(right)
print("total:", type(total), repr(total))
```

## 6. Errors, `try` and `except`
This program should add up all prices and print `Total: 15` – but the loop has **two bugs**.

### 🧪 Try it
1. Find both bugs **without** running the code.
2. Predict what the program prints.

```python
prices = [2, 5, 8]
total = 0
i = 0
try:
    while i < len(prices):
        total += prices(i)
    print("Total:", total)
except Exception:
    print("Something went wrong!")
```

💬 **Discuss:** why is `print("Total:", total)` never reached?

With `except Exception as e:` we can also see **which** error happened:

```python
prices = [2, 5, 8]
total = 0
i = 0
try:
    while i < len(prices):
        total += prices(i)
    print("Total:", total)
except Exception as e:
    print("Something went wrong:", type(e).__name__, "-", e)
```

### Fixing only the first bug
`prices[i]` uses **square brackets** to get an element. But `i` still never changes...

*(The `steps` counter and `break` are only a safety stop for the demo, so the notebook doesn't freeze.)*

```python
prices = [2, 5, 8]
total = 0
i = 0
steps = 0
while i < len(prices):
    total += prices[i]
    steps += 1
    if steps > 10:      # safety stop - not part of the real program!
        print("Still looping after 10 steps: infinite loop!")
        break
print("i is still", i)
```

```python
# Both bugs fixed
prices = [2, 5, 8]
total = 0
i = 0
while i < len(prices):
    total += prices[i]
    i += 1
print("Total:", total)
```

## 7. Expressions and Conditionals Together
Trace each part on the board: **which branch runs**, and what are the values afterwards?

```python
# Part 1
m = 9
n = 4
k = m % n
if k == 1:
    m = m + 5
    n -= 1
else:
    m = m * 2
print("after part 1:  m =", m, " n =", n, " k =", k)
```

```python
# Part 2
d = m // n
if d > 4 and n < 5:
    k = d - n
elif d == 4 or m < 10:
    k = m % d
else:
    k = 0
print("after part 2:  d =", d, " k =", k)
```

```python
# Part 3
m -= k * n
if m > n:
    if m % 2 == 0:
        n = n ** 2
    d = n - m
else:
    d = -1
print("after part 3:  m =", m, " n =", n, " d =", d)
```

```python
# Part 4
result = m + n + k + d
print("result =", result)
```

## 8. `for` Loops
`range(start, stop)` counts from `start` up to – but **not including** – `stop`.

```python
print(list(range(1, 6)))
print(list(range(5)))
print(list(range(0, 10, 3)))
```

### 🧪 Try it – write down every printed line first

```python
score = 10
for k in range(1, 6):
    if k % 3 == 0:
        score -= k
    else:
        score += 2
    print(k, score)
print("final:", score)
```

💬 **Discuss:** what changes if line 2 is `for k in range(6):`? Predict, then try it below.

```python
score = 10
for k in range(6):
    if k % 3 == 0:
        score -= k
    else:
        score += 2
    print(k, score)
print("final:", score)
```

## 9. `while` Loops
A `while` loop runs **as long as** its condition is `True`.

### 🧪 Try it – write down every printed line, and count the loop runs

```python
n = 50
while n > 5:
    if n % 2 == 0:
        n = n // 2
    else:
        n = n - 5
    print(n)
print("end")
```

💬 **Discuss:** what happens if the condition is `while n != 1:`? Will `n` ever be exactly 1?

*(Again with a safety stop so the notebook doesn't freeze.)*

```python
n = 50
steps = 0
while n != 1:
    if n % 2 == 0:
        n = n // 2
    else:
        n = n - 5
    print(n)
    steps += 1
    if steps > 8:       # safety stop - not part of the real program!
        print("... this never ends: infinite loop!")
        break
```

- Bug 1: `prices(i)` uses round brackets – that tries to *call* the list like a function → `TypeError`. It should be `prices[i]`.
- Bug 2: `i` is never increased → infinite loop. Add `i += 1`.
- The program prints `Something went wrong!`: the error jumps straight to `except` and skips the rest of the `try` block.

### Section 7 – expressions and conditionals

- Part 2: `d = 4`; first test is `False`, `elif` is `True` → `k = 14 % 4 = 2`
- Part 3: `m = 14 - 2 * 3 = 8`; both tests `True` → `n = 9`, `d = 1`
- Part 4: `result = 8 + 9 + 2 + 1 = 20`

### Section 8 – for loop
`1 12`, `2 14`, `3 11`, `4 13`, `5 15`, `final: 15`.
With `range(6)` there is one extra first line `0 10` (`0 % 3 == 0`, and subtracting 0 changes nothing).

### Section 9 – while loop
`25`, `20`, `10`, `5`, `end` – the loop body runs **4** times.
With `while n != 1:`, `n` goes 5 → 0 → 0 → 0 … and never becomes 1: infinite loop.
