## 1. Control Structures
Elements of imperative programming: expressions, variables, assignments and **control structures** (branches, **loops**).

- **Conditional statements** execute a (block of) statement(s) only **under certain conditions** → `if … else …`
- **Loop statements** execute a (block of) statement(s) **repeatedly** → `while …`, `for …`

## 2. Loops: `while`
```python
while expr:
    block
stmt
```
- Evaluate `expr`
- If the value is (or counts as) `True`:
  - execute the statements of `block`
  - then go back to the beginning of the `while` loop and evaluate `expr` again
- If the value is (or counts as) `False`:
  - continue with the next statement after the `while` block (`stmt`, on the same indent level as `while`)

````python
# Safe version: the extra test stops it after 10 rounds
a = 1
rounds = 0
while a < 5 and rounds < 10:
    print("Hi", a)
    a = a - 1
    rounds += 1
print("Stopped – without the rounds counter this would never end!")
````

## 6. `for` Loops and `range`
`for` loops iterate over all elements of collections, e.g. lists:
```python
for ele in some_list:
    print(ele)
```
We often want to iterate over all numbers in a certain range:
- `range(start, end)` behaves like the corresponding list
- The **end point is not included**: `range(0, 5)` → `[0, 1, 2, 3, 4]`
- Actually `range` evaluates to an **iterator**, not a list → use `list(range(0, 5))` to get a proper list (e.g., for printing)

````python
some_list = ["a", "b", "c"]
for ele in some_list:
    print(ele)
````

### `range(start, end, step)`
- `range(0, 10, 1)` → `[0, 1, 2, ..., 9]`
- `range(0, 10, 2)` → `[0, 2, 4, 6, 8]`
- `range(10, 0, -2)` → `[10, 8, 6, 4, 2]`

Short forms:
- `range(0, 10)` = `range(0, 10, 1)` → default step is 1
- `range(10)` = `range(0, 10, 1)` → default start is 0

````python
print(list(range(0, 10, 1)))
print(list(range(0, 10, 2)))
print(list(range(10, 0, -2)))
print(list(range(0, 10)))
print(list(range(10)))
````

## 10. `break`, again: `else` on Loops
It's sometimes convenient to know whether a loop was terminated by executing a `break` statement or not.

Loops can have an optional `else` block at the end:
- The `else` block is executed when the loop terminates "properly",
  i.e. **unless** the loop was terminated by executing a `break` statement

````python
some_list_of_numbers = [2, 4, 8, 6, 3, 2, 4]

for n in some_list_of_numbers:
    if n % 2 == 1:
        print("some numbers are odd")
        break
else:
    print("all numbers are even")
````
