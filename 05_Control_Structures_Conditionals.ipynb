## 2. Conditions: `if … else …`
```python
if expr1:
    block1
else:
    block2
```
- If `expr1` evaluates to a value that counts as `True` → run `block1`, otherwise → run `block2`
- The `else` part is **optional**
- Implicit conversion: `0`, `''`, `[]` → `False`; all other values → `True`
- Blocks are marked by **indentation**

**Example:** read in an integer and test whether it is even or odd.

````python
# Version 1 – can you improve this?
num = int(input("Enter an integer: "))

if num % 2 == 0:
    print(f"{num} is even.")

if num % 2 != 0:
    print(f"{num} is odd.")
````

````python
# Version 2 – the test expression is only evaluated once → more efficient
num = int(input("Enter an integer: "))

if num % 2 == 0:
    print(f"{num} is even.")
else:
    print(f"{num} is odd.")
````

## 3. Blocks and Indentation
A **block** groups statements. Statements of the same block
- must have the **same indentation**
- must use the **same type of whitespace** (tabs or spaces)

Best practices: never mix spaces and tabs · use 4 spaces · let VS Code do the indenting for you.
Blocks can be **nested**.

````python
a, b = 1, 2
if a < b:
    print("foo")
    a += 1
else:
    print("bar")
    b -= 1
print(a, b)
````

````python
# Nested block
a, b = 3, 5
if a < b:
    print("foo")
    a += 1
    if a > 3:
        print("bar")
````

## 4. Conditions: `if … elif … else`
```python
if expr1:
    block1
elif expr2:
    block2
elif ...:
    ...
else:
    blockn
```
- Runs the **first** block whose test is `True` – all other blocks are skipped
- `else` is optional
- `elif` is the same as nested `if … else`, but much more readable

````python
# With elif …
x = 0
if x > 0:
    print("positive")
elif x < 0:
    print("negative")
else:
    print("zero")
````

````python
# … is the same as nested if/else
x = 0
if x > 0:
    print("positive")
else:
    if x < 0:
        print("negative")
    else:
        print("zero")
````

## 6. Conditional Expressions
Instead of a four-line `if … else` assignment, you can write one line:
```python
m = a if a > b else b
```
The right-hand side is an **expression**, so it can be part of longer expressions.

````python
a, b = 4, 9

# long form
if a > b:
    m = a
else:
    m = b
print(m)

# conditional expression
m = a if a > b else b
print(m)
````

````python
a, b = 4, -2
m = (a if a > b else b) + (7 if b < 0 else 11)
print(m)
````

## 7. Match Statements (Python 3.10+)
```python
match term:
    case pattern-1:
        block1
    case pattern-2 | pattern-3:
        block2
    case _:
        default-block
```
- Tests `term` against each pattern **in order** until the first match
- Uses `==` comparison – ⚠️ **not** a full Boolean expression (no `case > 90`)
- Combine patterns with `|`
- `case _` catches everything else

````python
status = 418       # in the slides: status = get_http_status()
match status:
    case 400:
        print("Bad request")
    case 404:
        print("Not found")
    case 418:
        print("I'm a teapot")
    case _:
        print("Something's wrong with the internet")
````
